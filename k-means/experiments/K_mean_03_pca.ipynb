# KMeans 개선안 3: PCA 차원 축소

10,000차원 픽셀 벡터를 PCA로 압축한 뒤 KMeans를 수행합니다. 노이즈를 줄이고 학습 속도를 개선할 수 있습니다.

```python
from pathlib import Path
from urllib.request import urlretrieve

data_path = Path('fruits_300.npy')
if not data_path.exists():
    urlretrieve('https://bit.ly/fruits_300_data', data_path)
```

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

fruits = np.load('fruits_300.npy')
fruits_2d = fruits.reshape(-1, 100 * 100)

pca = PCA(n_components=50, random_state=42)
fruits_pca = pca.fit_transform(fruits_2d)

print('PCA shape:', fruits_pca.shape)
print('explained variance ratio sum:', pca.explained_variance_ratio_.sum())
```

```python
km = KMeans(n_clusters=3, n_init=20, random_state=42)
km.fit(fruits_pca)

print('labels:', np.unique(km.labels_, return_counts=True))
print('inertia:', km.inertia_)
print('n_iter:', km.n_iter_)
```

```python
for cluster_id in range(3):
    print(f'cluster {cluster_id}:', np.where(km.labels_ == cluster_id)[0][:20])
```
